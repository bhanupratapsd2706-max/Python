Grid() = Helps make plots easier to read by adding reference lines.

```python
X = [1,2,3,4,5]
Y = [5,10,15,20,25]
plt.grid(linewidth=2,
         color="lightgray",
         linestyle="dashed"
         )
plt.plot(X,Y)
plt.show()
```

Bar chart = Compare Categories of data by representing each category with a bar

```python
import matplotlib.pyplot as plt
import numpy as np
Categories = np.array(["grain","Fruits","Vegies","Sweets","Dairy"])
Values = np.array([4,2,5,7,2])
plt.bar(Categories,Values)
plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```

Bar Chart = Circular chart divided into Slices to show percentages of the total. Good for visualizing distribution among categories.

```python
import matplotlib.pyplot as plt
import numpy as np
Catogries = ["Freshmen", "Sophomores","Juniour","Seniors"]
Values = np.array([300,250,275,255])
colors = ["red","blue","Grey","green"]
plt.pie(Values,labels=Catogries,
        autopct="%1.1f%%",
        colors=colors,
        explode=(0,0,0,0.1),
        shadow=True,
        startangle=90)
plt.title("Bhanu College")
plt.show()


```

Scatter Graph = Shows the relationship between two variables helps to identify a correlation (+,-,None) .
Example : Study hours vs . Test scores

```python
import matplotlib.pyplot as plt
import numpy as np
X = [0,11,2,3,4,5,6,7,7,8]
y = [55,60,65,62,68,70,75,78,82,85]
plt.scatter(X,y)
plt.title("Hours vs scores")
plt.xlabel("Hours")
plt.ylabel("scores")
plt.show()
```

Histogram = A Visual representation of the distribution of quantitative data.
they group values into bins ( intervals)
and counts how many fall in each range .

```python
import matplotlib.pyplot as plt
import numpy as np
scores = np.random.normal(loc=80,scale=10,size=100)
scores = np.clip(scores,0,100)
plt.hist(scores,bins=10,
         edgecolor="black")
plt.show
```

Figure = The entire canvas


Ax = A single plot (subplot)

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1,2,3,4,5])
Figure , axes = plt.subplots(2,2)
axes[0,0].plot(x,x*2,color="red")
axes[0,0].set_title("x*2")
axes[0,1].plot(x,x*2,color="blue")
axes[0,1].set_title("x*3")
axes[1,0].plot(x,x*2,color="green")
axes[1,0].set_title("x*4")
axes[1,1].plot(x,x*2,color="grey")
axes[1,1].set_title("x*5")
plt.show()
```
