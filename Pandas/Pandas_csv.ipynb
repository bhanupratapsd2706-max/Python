DataFrames = A tabular data structure with rows and column (2-Dimensional).Similar to an excel spreadsheet.

```python
import pandas as pd
data = {
    "Name":["Bhanu","Prabhat","Saksham"],
    "Age":[20,21,24]
}
df = pd.DataFrame(data,index=["Employee1","Employee2","Employee3"])
print(df)
```

How to add a new Row and Column

```python
import pandas as pd
data = {
    "Name":["Bhanu","Prabhat","Saksham"],
    "Age":[20,21,24]
}
df = pd.DataFrame(data,index=["Employee1","Employee2","Employee3"])

#Add a new Column
df["Job"] = ["Cook","N/A","Cashier"]

# Add a New Row
new_row = pd.DataFrame([{"Name":"Sandy","Age":34,"Job":"Engineer"},
                         {"Name":"Rana","Age":28,"Job":"Advocate"}],
                        index=["Employee 4","Employee 5"])
df = pd.concat([df, new_row])
print(df)
```

```python
import pandas as pd
df = pd.read_csv("placement.csv")
# for average package and cgpa
print(df.mean(numeric_only=True))
# sum of cgpa and package
print(df.sum(numeric_only=True))
# to find minimun cgpa and package
print(df.min(numeric_only=True))
# to find maximun cgpa and package
print(df.max(numeric_only=True))

print(df.count())
```
